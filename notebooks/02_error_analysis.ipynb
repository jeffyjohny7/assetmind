```python
from pathlib import Path
import pandas as pd
import plotly.express as px

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
p = pd.read_parquet(ROOT / "reports/predictions_FD001.parquet")
p["error"] = p["pred_lightgbm"] - p["true_rul"]   # positive = late (dangerous)
p["abs_error"] = p["error"].abs()
```

```python
fig = px.scatter(p, x="true_rul", y="pred_lightgbm", hover_data=["unit", "n_cycles"],
                 title="LightGBM: predicted vs true RUL")
fig.add_shape(type="line", x0=0, y0=0, x1=150, y1=150, line=dict(dash="dash"))
fig.show()
```

```python
px.scatter(p, x="n_cycles", y="error", hover_data=["unit", "true_rul"],
           title="Error vs how much history the engine had").show()
```

```python
print("Late predictions (dangerous):", int((p["error"] > 0).sum()))
print("Early predictions (safe):    ", int((p["error"] < 0).sum()))
p.sort_values("abs_error", ascending=False).head(10)
```
